# Exploration of Lineup Data


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

lineups = pd.read_csv("../data/processed/lineups_clean.csv")

lineups.sample(5)
```

```python
lineup_counts = (
    lineups
    .groupby(["festival", "year"])
    .size()
    .reset_index(name="artist_count")
)

sns.lineplot(
    data=lineup_counts,
    x="year",
    y="artist_count",
    hue="festival",
    marker="o"
)

plt.title("Festival Lineup Size Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Artists")
plt.show()
```
